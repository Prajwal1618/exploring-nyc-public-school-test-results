# Exploring NYC Public School Test Result Scores

This project analyzes SAT performance across New York City public schools using data on math, reading, and writing scores.

The analysis focuses on three questions:

1. Which NYC schools have the best math results?
2. Which schools have the highest combined SAT scores?
3. Which NYC borough has the largest variation in combined SAT scores?

## 1. Load and Explore the Data

First, import pandas and load the NYC public school SAT dataset.

```python
import pandas as pd

schools = pd.read_csv("schools.csv")
schools.head()
```

```python
schools.info()
```

## 2. Schools with the Best Math Results

The maximum possible SAT Math score is 800.  
Schools with an average math score of at least 80% of the maximum score are considered to have the best math results.

Therefore, the minimum qualifying score is 640.

```python
best_math_schools = (
    schools[schools["average_math"] >= 0.8 * 800]
    [["school_name", "average_math"]]
    .sort_values("average_math", ascending=False)
)
```

```python
best_math_schools
```

The results show the schools with an average math score of at least 640, ranked from highest to lowest.

## 3. Top 10 Schools Based on Combined SAT Scores

To compare overall SAT performance, I calculate a combined SAT score by adding the average math, reading, and writing scores for each school.

The schools are then sorted by their combined SAT score, and the top 10 schools are selected.

```python
schools_new = schools.copy()

schools_new["total_SAT"] = (
    schools_new["average_math"]
    + schools_new["average_reading"]
    + schools_new["average_writing"]
)
```

```python
schools_new.head()
```

```python
top_10_schools = (
    schools_new[["school_name", "total_SAT"]]
    .sort_values("total_SAT", ascending=False)
    .head(10)
)
```

```python
top_10_schools
```

These are the 10 schools with the highest combined average SAT scores across math, reading, and writing.

## 4. Borough with the Largest Variation in SAT Scores

Standard deviation measures how spread out the combined SAT scores are within each borough.

I will:
1. Group schools by borough.
2. Calculate the number of schools, average SAT score, and standard deviation of SAT scores.
3. Identify the borough with the largest standard deviation.
4. Round the numerical results to two decimal places.

```python
schools_grpby_borough = (
    schools_new
    .groupby("borough")["total_SAT"]
    .agg(["count", "mean", "std"])
)

schools_grpby_borough
```

```python
largest_std_dev = (
    schools_grpby_borough[
        schools_grpby_borough["std"] == schools_grpby_borough["std"].max()
    ]
)

largest_std_dev
```

```python
largest_std_dev = (
    largest_std_dev
    .rename(
        columns={
            "borough": "borough",
            "count": "num_schools",
            "mean": "average_SAT",
            "std": "std_SAT"
        }
    )
    .round(2)
)
```

```python
largest_std_dev
```

### Finding

Manhattan has the largest standard deviation in combined SAT scores, with a standard deviation of 230.29 across 89 schools.

## Conclusion

The analysis identified the NYC schools with the strongest math performance and the top 10 schools based on combined SAT scores. Manhattan had the largest variation in combined SAT performance, with a standard deviation of 230.29 across 89 schools.
